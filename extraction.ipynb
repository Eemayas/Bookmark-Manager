```python
from bs4 import BeautifulSoup
import json

def parse_bookmarks(html_file):
    try:
        with open(html_file, 'r', encoding='utf-8') as file:
            html_content = file.read()
    except FileNotFoundError:
        print(f"File {html_file} not found.")
        return []

    soup = BeautifulSoup(html_content, 'html.parser')

    def process_dl(dl_tag, current_path):
        bookmarks = []
        current_folder = None
        
        all_tags = dl_tag.find_all(['dt', 'dl'], recursive=False)
        
        if len(all_tags) == 0:
            all_tags = dl_tag.find_all(['dt', 'dl'])
        
        for tag in all_tags:
            if tag.name == 'dt':
                h3_tag = tag.find('h3')
                a_tag = tag.find('a')
                if h3_tag:  
                    current_folder = h3_tag.text
                    nested_dl = tag.find_next_sibling('dl')
                    if nested_dl:
                        new_path = current_path + [current_folder]
                        bookmarks.extend(process_dl(nested_dl, new_path))
                elif a_tag:  
                    link = a_tag.get('href')
                    name = a_tag.text
                    path = " / ".join(current_path)
                    bookmarks.append({
                        'name': name,
                        'link': link,
                        'path': path
                    })
            elif tag.name == 'dl' and current_folder:  
                new_path = current_path + [current_folder]
                bookmarks.extend(process_dl(tag, new_path))
        
        return bookmarks

    dl_tag = soup.find('dl')
    if dl_tag is None:
        print("No bookmarks found in the HTML file.")
        return []
    
    bookmarks = process_dl(dl_tag, [])
    
    return bookmarks

html_file = 'bookmarks_10_13_24.html'
parsed_bookmarks = parse_bookmarks(html_file)

if parsed_bookmarks:
    print(f"Number of parsed bookmarks: {len(parsed_bookmarks)}")
    # remove the duplicate in paresed bookmarks
    bookmark_set = set()
    bookmark_list = []

    for i, bookmark in enumerate(parsed_bookmarks):
        if bookmark['link'] not in bookmark_set:
            bookmark_set.add(bookmark['link'])
            bookmark_list.append({
                "id": len(bookmark_list) + 1,
                "name": bookmark['name'],
                "url": bookmark['link'],
                "description": "No description",
                "tags": ["bookmarks"],
                "categories": bookmark['path']
            })
    output_file = 'constants/bookmarks.json'
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(bookmark_list, f, ensure_ascii=False, indent=2)
    print(f"Bookmarks saved to {output_file}")
    print(f"Number of bookmarks: {len(bookmark_list)}")
```

```python
from bs4 import BeautifulSoup


def transform_string(s):
    # Split the string by underscores
    parts = s.split("_")
    # Capitalize the first letter of each part
    capitalized_parts = [part.capitalize() for part in parts]
    # Join the parts with a space
    return " ".join(capitalized_parts)


# List of HTML contents and their corresponding categories
htmls = [
    ("stock_photos", stock_photos_html_content),
    ("stock_videos", stock_video_html_content),
    ("icons", icons_html_content),
    ("icon_finder", icon_finder_html_content),
    ("illustrations", illustrations_html_content),
    ("font_tools", font_tools_html_content),
    ("gradient_tools", gradient_tools_html_content),
    ("image_tools", image_tools_html_content),
    ("builder", builder_html_content),
    ("collaboration_tools", collaboration_tools_html_content),
    ("design_systems", design_systems_html_content),
    ("designer_podcasts", designer_podcasts_html_content),
    ("ux_research", ux_research_html_content),
    ("productivity", productivity_html_content),
    ("inspiration", inspiration_html_content),
    ("instagram_tools", instagram_tools_html_content),
    ("logo_tools", logo_tools_html_content),
    ("mockup_tools", mockup_tools_html_content),
    ("placeholder_texts", placeholder_texts_html_content),
    ("prototyping", prototyping_html_content),
    ("wireframe_tools", wireframe_tools_html_content),
    ("animation_tools", animation_tools_html_content),
    ("social_media_tools", social_media_tools_html_content),
    ("sound_and_audio", sound_and_audio_html_content),
    ("user_flow", user_flow_html_content),
    ("design_tools", design_tools_html_content),
    ("notion_tools", notion_tools_html_content),
    ("generators", generators_html_content),
    ("job_boards", job_boards_html_content),
    ("handoff", handoff_html_content),
]

popular_links = {}

for category, html_content in htmls:
    # Parse the HTML
    soup = BeautifulSoup(html_content, "html.parser")

    # Find all 'a' tags with the class 'rsli-item'
    a_tags = soup.find_all("a", class_="rsli-item")

    # Create an array of dictionaries
    temp_results = []
    for a_tag in a_tags:
        title_tag = a_tag.find("h3", class_="rsli-title")
        description = title_tag.get("title", "") if title_tag else ""
        name = title_tag.text.strip() if title_tag else ""
        url = a_tag.get("href", "")
        temp_results.append(
            {
                "id": len(temp_results) + 1,
                "description": description,
                "name": name,
                "url": url,
                "tags": transform_string(category),
                "categories": f"/{transform_string(category)}",
            }
        )

    # Assign the results to the category in the popular_links dictionary
    popular_links[category] = temp_results

# popular_links now contains parsed results for each category
output_file = "constants/popularslinks.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(popular_links, f, ensure_ascii=False, indent=2)
print(f"Popular Links saved to {popular_links}")
print(f"Number of Popular Links: {len(popular_links)}")
```
