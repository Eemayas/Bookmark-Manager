```python
from bs4 import BeautifulSoup

# Read the HTML file
with open('bookmarks_10_13_24.html', 'r', encoding='utf-8') as file:
    html_content = file.read()

# Parse the HTML content
soup = BeautifulSoup(html_content, 'html.parser')

# Extract all the <a> tags
links = soup.find_all('a')

# Iterate over each link and print the name and href
for link in links:
    name = link.text
    href = link.get('href')
    print(f'Name: {name}, Link: {href}')
```

```python
def parse_bookmarks(html_file):
    with open(html_file, 'r', encoding='utf-8') as file:
        html_content = file.read()

    soup = BeautifulSoup(html_content, 'html.parser')

    def process_dl(dl_tag, current_path):
        bookmarks = []
        current_folder = None
        
        # print(f"Processing dl_tag: {dl_tag}")  
        # print(f"DL tag contents: {dl_tag.contents}")  
        all_tags = dl_tag.find_all(['dt', 'dl'], recursive=False)
        # print(f"Number of tags found: {len(all_tags)}")  
        
        # If no tags found, try without recursive=False
        if len(all_tags) == 0:
            all_tags = dl_tag.find_all(['dt', 'dl'])
            # print(f"Number of tags found (recursive): {len(all_tags)}")
        
        for tag in all_tags:
            # print(f"Processing tag: {tag.name}")  
            if tag.name == 'dt':
                h3_tag = tag.find('h3')
                a_tag = tag.find('a')
                if h3_tag:  # It's a folder
                    current_folder = h3_tag.text
                    # print(f"Found folder: {current_folder}") 
                    nested_dl = tag.find_next_sibling('dl')
                    if nested_dl:
                        new_path = current_path + [current_folder]
                        bookmarks.extend(process_dl(nested_dl, new_path))
                elif a_tag:  # It's a bookmark
                    link = a_tag.get('href')
                    name = a_tag.text
                    path = " / ".join(current_path)
                    # print(f"Found bookmark: {name}")  
                    bookmarks.append({
                        'name': name,
                        'link': link,
                        'path': path
                    })
            elif tag.name == 'dl' and current_folder:  # Nested <dl> after an <h3>
                new_path = current_path + [current_folder]
                bookmarks.extend(process_dl(tag, new_path))
        
        return bookmarks

    # Start processing the top-level DL tag
    dl_tag = soup.find('dl')
    # print(f"Top-level DL tag: {dl_tag}")  
    # print(f"HTML structure around DL tag: {dl_tag.parent}")  
    bookmarks = process_dl(dl_tag, [])
    
    return bookmarks
# Path to your bookmarks HTML file
html_file = 'bookmarks_10_13_24.html'
parsed_bookmarks = parse_bookmarks(html_file)

# Print the number of parsed bookmarks
print(f"Number of parsed bookmarks: {len(parsed_bookmarks)}")

# Create a list of bookmark objects
bookmark_list = [
    {
        "id": i + 1,
        "name": bookmark['name'],
        "url": bookmark['link'],
        "description": "No description",
        "tags": ["bookmarks"],
        "categories": bookmark['path']
    }
    for i, bookmark in enumerate(parsed_bookmarks)
]
# Save bookmark_list to a JSON file
import json

output_file = 'constants/bookmarks.json'

with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(bookmark_list, f, ensure_ascii=False, indent=2)

print(f"Bookmarks saved to {output_file}")


# Optionally, you can print the length of the list to see how many bookmarks were processed
print(f"Number of bookmarks: {len(bookmark_list)}")

# If you want to see the first few items in the list, you can do:
print("First 5 bookmarks:")
for bookmark in bookmark_list[:5]:
    print(bookmark)
```
